```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
import numpy as np

# Circular mask
x, y = np.ogrid[:500, :500]
mask = ((x - 250)**2)/0.7 + (y - 250)**2 > 300**2
mask = 255 * mask.astype(int)

topics = [
    "Data Augmentation", "Model Architecture", "Classification Head", "Loss function",
    "Model Depth", "Decay Scheduling", "Optimizer", "Convolution Type", "Weight Decay",
    "Feature Aggregation", "Padding", "Initialization", "Monte Carlo Dropout", "Dropout", "Stochastic Depth", 
    "Batch Norm", "Learning Rate", "Gradient Clipping", "Feature Maps", "Grad CAM",
]
weights = list(np.sqrt(1 + np.arange(len(topics), 0, -1)))
word_weights = dict(zip(topics, weights))

wordcloud = WordCloud(
    width=1800,
    height=1000,
    mask=mask,
    background_color='white',
    colormap='plasma',
    prefer_horizontal=0.1,
    max_font_size=50,
    min_font_size=15,
    random_state=5
).generate_from_frequencies(word_weights)

plt.figure(figsize=(18, 10))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.title("Word Cloud – Styled", fontsize=18, color='white')
plt.tight_layout(pad=0)
plt.show()
```
